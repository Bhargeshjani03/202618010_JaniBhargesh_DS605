Task 2 - Vectorized Arithmetic and Linear Algebra

Step 1: Create two matrices and do addition, element-wise multiplication, matrix multiplication

```python
import numpy as np

M1 = np.array([[1, 2, 3],
               [4, 5, 6]])

M2 = np.array([[7, 8, 9],
               [10, 11, 12]])
add_result = M1 + M2
print("Addition:\n", add_result)
elementwise_mult = M1 * M2
print("Element-wise multiplication:\n", elementwise_mult)

M3 = np.array([[1, 2],
               [3, 4],
               [5, 6]])

matmul_result = M1 @ M3 
print("Matrix multiplication (M1 @ M3):\n", matmul_result)
```

Step 2: Square matrix — transpose and determinant

```python

S = np.array([[4, 7, 2],
              [3, 6, 1],
              [2, 5, 3]])


S_transpose = S.T 
print("Original S:\n", S)
print("Transpose:\n", S_transpose)

# Determinant
det_S = np.linalg.det(S)
print("Determinant:", round(det_S,2))
```

Step 3: Inverse (if invertible) + verify with np.allclose()

```python
if det_S != 0:
    S_inv = np.linalg.inv(S)
    print("Inverse:\n", S_inv)

    identity_check = S @ S_inv
    print("S @ S_inv:\n", identity_check)

    is_identity = np.allclose(identity_check, np.eye(3))
    print("Is S @ S_inv approximately the identity matrix?", is_identity)
else:
    print("Matrix S is singular (determinant = 0), inverse does not exist.")
```

Task 3 - Normal Distribution and Histogram

Step 1: Generate the samples

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)  


mu = 50      
sigma = 10    
n = 1000     

data = np.random.normal(loc=mu, scale=sigma, size=n)

print("First 10 values:", data[:10])
print("Number of samples generated:", len(data))
```

We can see the data is created around the population mean

Step 2: Sample mean and sample standard deviation

```python
sample_mean = np.mean(data)
sample_std = np.std(data)

print(f"Chosen population mean (mu): {mu}")
print(f"Sample mean: {sample_mean:.4f}")
print(f"Difference: {abs(mu - sample_mean):.4f}")

print(f"\nChosen population std (sigma): {sigma}")
print(f"Sample std: {sample_std:.4f}")
print(f"Difference: {abs(sigma - sample_std):.4f}")
```

We can see that sample mean and standard deviation nearly matches population mean and standard deviation

Step 3: Plot the histogram

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

mu = 50
sigma = 10
n = 1000

data = np.random.normal(loc=mu, scale=sigma, size=n)

sample_mean = np.mean(data)
sample_std = np.std(data)

print(f"Chosen mean: {mu}, Sample mean: {sample_mean:.4f}")
print(f"Chosen std: {sigma}, Sample std: {sample_std:.4f}")

plt.figure(figsize=(8, 5))
plt.hist(data, bins=30, color='steelblue', edgecolor='black', alpha=0.7)
plt.axvline(sample_mean, color='red', linestyle='--', linewidth=2, label=f'Sample mean = {sample_mean:.2f}')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title(f'Histogram of {n} samples from Normal(μ={mu}, σ={sigma})')
plt.legend()
plt.show()
```

Task 5 - Filtering and Querying

Query 1: Male passengers older than 50

```python

male_over_50 = df[(df['Sex'] == 'male') & (df['Age'] > 50)]
print("Male passengers older than 50:", len(male_over_50))

# Equivalent using query()
male_over_50_q = df.query("Sex == 'male' and Age > 50")
print("Using query():", len(male_over_50_q))
```

Query 2: Female first-class passengers — count and survival %

```python
female_1st = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]
count = len(female_1st)
survival_pct = female_1st['Survived'].mean() * 100

print(f"Female first-class passengers: {count}")
print(f"Survival percentage: {survival_pct:.2f}%")
```

Query 3: Age 20–40, Fare above overall median, and survived

```python
fare_median = df['Fare'].median()

condition3 = (df['Age'] >= 20) & (df['Age'] <= 40) & (df['Fare'] > fare_median) & (df['Survived'] == 1)
result3 = df[condition3]

print("Fare median used:", fare_median)
print("Count (age 20-40, fare > median, survived):", len(result3))
```

Query 4: Traveling alone, age < 30, did not survive

```python
condition4 = (df['SibSp'] == 0) & (df['Parch'] == 0) & (df['Age'] < 30) & (df['Survived'] == 0)
result4 = df[condition4]

print("Count (alone, age < 30, did not survive):", len(result4))
```

Query 5: Embarked='S', Pclass 2 or 3, Fare above Southampton median

```python

southampton_df = df[df['Embarked'] == 'S']
southampton_fare_median = southampton_df['Fare'].median()
condition5 = (df['Embarked'] == 'S') & (df['Pclass'].isin([2, 3])) & (df['Fare'] > southampton_fare_median)
result5 = df[condition5]

print("Southampton fare median:", southampton_fare_median)
print("Count (Embarked=S, Pclass 2 or 3, Fare > Southampton median):", len(result5))
```

Task 7 - Missing Values and Fare Outliers

Step 1: Missing value count and percentage per column

```python
import pandas as pd
import numpy as np
df = pd.read_csv('train.csv')
missing_count = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing %': missing_pct.round(2)
})

missing_summary = missing_summary[missing_summary['Missing Count'] > 0].sort_values('Missing Count', ascending=False)
print(missing_summary)
```

Step 2: Plot missing counts as a bar chart

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))
missing_summary['Missing Count'].plot(kind='bar', color='steelblue', edgecolor='black')
plt.ylabel('Missing Count')
plt.title('Missing Values per Column')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Step 3: Fill missing Age with mean — before/after comparison

```python
print("Missing Age BEFORE filling:", df['Age'].isnull().sum())

df['Age_filled_mean'] = df['Age'].fillna(df['Age'].mean())

print("Missing Age AFTER filling:", df['Age_filled_mean'].isnull().sum())
```

Step 4: Try multiple imputation strategies (mean, median, mode, random)

```python
np.random.seed(42)

df['Age_mean'] = df['Age'].fillna(df['Age'].mean())

df['Age_median'] = df['Age'].fillna(df['Age'].median())

df['Age_mode'] = df['Age'].fillna(df['Age'].mode()[0])

known_ages = df['Age'].dropna()
random_fill = np.random.choice(known_ages, size=df['Age'].isnull().sum())

df['Age_random'] = df['Age'].copy()
df.loc[df['Age_random'].isnull(), 'Age_random'] = random_fill

print(df[['Age', 'Age_mean', 'Age_median', 'Age_mode', 'Age_random']].isnull().sum())
print("\nSummary comparison of means after imputation:")
print(df[['Age_mean', 'Age_median', 'Age_mode', 'Age_random']].mean())
```

Step 5: Fare — Q1, Q3, IQR, outlier bounds and count

```python
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Fare'] < lower_bound) | (df['Fare'] > upper_bound)]

print(f"Q1: {Q1}")
print(f"Q3: {Q3}")
print(f"IQR: {IQR}")
print(f"Lower bound: {lower_bound}")
print(f"Upper bound: {upper_bound}")
print(f"Number of Fare outliers: {len(outliers)}")
```

Task 8 - Features and Pivot Table

Step 1: Create FamilySize and IsAlone

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)

print(df[['SibSp', 'Parch', 'FamilySize', 'IsAlone']].head(10))
```

Step 2: Pivot table — rows=Sex, columns=Pclass, values=mean Survived

```python
pivot = df.pivot_table(index='Sex', columns='Pclass', values='Survived', aggfunc='mean')
print(pivot)
```

Step 3: Find highest and lowest groups programmatically

```python
# Flatten to find the single highest and lowest (Sex, Pclass) combination
stacked = pivot.stack()

highest = stacked.idxmax()
highest_val = stacked.max()

lowest = stacked.idxmin()
lowest_val = stacked.min()

print(f"Highest survival rate: Sex={highest[0]}, Pclass={highest[1]} → {highest_val:.4f}")
print(f"Lowest survival rate: Sex={lowest[0]}, Pclass={lowest[1]} → {lowest_val:.4f}")
```

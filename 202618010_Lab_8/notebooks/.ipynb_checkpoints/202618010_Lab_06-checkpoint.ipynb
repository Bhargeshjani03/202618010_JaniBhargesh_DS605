```python
import os
from pathlib import Path

import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
IMAGE_DIR = Path("../data/images")

CRACK_DIR = IMAGE_DIR / "Cracks"
NON_CRACK_DIR = IMAGE_DIR / "NonCracks"

print("Cracks folder:", CRACK_DIR)
print("Non-Cracks folder:", NON_CRACK_DIR)
```

```python
image_extensions = {".jpg"}

crack_images = [
    p for p in CRACK_DIR.iterdir()
    if p.suffix.lower() in image_extensions
]

non_crack_images = [
    p for p in NON_CRACK_DIR.iterdir()
    if p.suffix.lower() in image_extensions
]

print("Number of crack images:", len(crack_images))
print("Number of non-crack images:", len(non_crack_images))
print("Total images:", len(crack_images) + len(non_crack_images))
```

```python
def inspect_images(image_paths, n=10):
    results = []

    for path in image_paths[:n]:
        img = cv2.imread(str(path))

        if img is None:
            print(f"Could not read: {path}")
            continue

        results.append({
            "image": path.name,
            "width": img.shape[1],
            "height": img.shape[0],
            "channels": 1 if len(img.shape) == 2 else img.shape[2]
        })

    return pd.DataFrame(results)
```

```python
crack_info = inspect_images(crack_images, n=200)
non_crack_info = inspect_images(non_crack_images, n=200)

print("Crack images:")
display(crack_info)

print("Non-crack images:")
display(non_crack_info)
```

```python
print("Crack image dimensions:")
print(crack_info[["width", "height", "channels"]].value_counts())

print("\nNon-crack image dimensions:")
print(non_crack_info[["width", "height", "channels"]].value_counts())
```

```python
IMG_SIZE = (224, 224)

def preprocess_image(image_path):
    # Read image
    image = cv2.imread(str(image_path))

    # Resize
    resized = cv2.resize(image, IMG_SIZE)

    # Convert BGR to grayscale
    gray = cv2.cvtColor(resized, cv2.COLOR_BGR2GRAY)

    return resized, gray
```

```python
sample_path = crack_images[0]

resized, gray = preprocess_image(sample_path)

print("Original shape:", cv2.imread(str(sample_path)).shape)
print("Resized shape:", resized.shape)
print("Grayscale shape:", gray.shape)
```

```python
def extract_features(image_path, label):
    # Read image
    image = cv2.imread(str(image_path))

    # Resize
    resized = cv2.resize(image, (224, 224))

    # Convert to grayscale
    gray = cv2.cvtColor(resized, cv2.COLOR_BGR2GRAY)

    # Intensity features
    mean_brightness = np.mean(gray)
    contrast = np.std(gray)

    dark_pixel_ratio = np.mean(gray < 50)
    bright_pixel_ratio = np.mean(gray > 200)

    min_intensity = np.min(gray)
    max_intensity = np.max(gray)
    median_intensity = np.median(gray)

    # Canny edge detection
    edges = cv2.Canny(gray, 100, 200)

    # Edge features
    edge_count = np.sum(edges > 0)
    edge_density = edge_count / edges.size

    return {
        "image_name": image_path.name,
        "label": label,
        "mean_brightness": mean_brightness,
        "contrast": contrast,
        "dark_pixel_ratio": dark_pixel_ratio,
        "bright_pixel_ratio": bright_pixel_ratio,
        "min_intensity": min_intensity,
        "max_intensity": max_intensity,
        "median_intensity": median_intensity,
        "edge_count": edge_count,
        "edge_density": edge_density
    }
```

```python
features = []

for image_path in crack_images:
    row = extract_features(image_path, label=1)
    features.append(row)

print("Crack images processed:", len(features))
```

```python
for image_path in non_crack_images:
    row = extract_features(image_path, label=0)
    features.append(row)

print("Total images processed:", len(features))
```

```python
image_features = pd.DataFrame(features)

print("Shape:", image_features.shape)

display(image_features.head())
```

```python
import matplotlib.pyplot as plt

features_to_plot = [
    "mean_brightness",
    "dark_pixel_ratio",
    "edge_density"
]

for feature in features_to_plot:

    plt.figure(figsize=(7, 5))

    image_features.boxplot(
        column=feature,
        by="label"
    )

    plt.title(f"{feature} by Class")
    plt.suptitle("")
    plt.xlabel("Class (0 = Non-Crack, 1 = Crack)")
    plt.ylabel(feature)

    plt.show()
```

```python
feature_columns = [
    "mean_brightness",
    "contrast",
    "dark_pixel_ratio",
    "bright_pixel_ratio",
    "min_intensity",
    "max_intensity",
    "median_intensity",
    "edge_density"
]

X = image_features[feature_columns]
y = image_features["label"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(random_state=42))
])
```

```python
import time

start_time = time.perf_counter()

logistic_model.fit(X_train, y_train)

training_time = time.perf_counter() - start_time

print("Training time:", training_time, "seconds")
```

```python
start_time = time.perf_counter()

y_pred = logistic_model.predict(X_test)

prediction_time = time.perf_counter() - start_time

print("Prediction time:", prediction_time, "seconds")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1-score :", f1)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 5))

plt.imshow(cm)

plt.title("Confusion Matrix - Logistic Regression")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")

plt.xticks([0, 1], ["Non-Crack", "Crack"])
plt.yticks([0, 1], ["Non-Crack", "Crack"])

for i in range(2):
    for j in range(2):
        plt.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center"
        )

plt.colorbar()
plt.tight_layout()
plt.show()
```

```python
image_results = []

image_results.append({
    "Model": "Logistic Regression",
    "Features": X.shape[1],
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "F1-Score": f1,
    "Training Time (s)": training_time,
    "Prediction Time (s)": prediction_time
})

results_df = pd.DataFrame(image_results)

display(results_df)
```
